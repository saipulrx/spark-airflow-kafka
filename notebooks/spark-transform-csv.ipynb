# Latihan Spark Transform data from csv files

## Import all library needed

```python
from pyspark.sql import SparkSession
import pyspark
from pyspark.sql import functions as F
from pyspark.sql.functions import avg, max, min, count, sum, desc
```

## Inisialisasi Spark

```python
sparkcontext = pyspark.SparkContext.getOrCreate(conf=(
        pyspark
        .SparkConf()
        .setAppName('RuangData')
        .setMaster('local')
    ))
sparkcontext.setLogLevel("WARN")

spark = pyspark.sql.SparkSession(sparkcontext.getOrCreate())
```

## Read data online retail dataset

```python
df_retail = spark.read.format("csv").option("header","true").csv("/resources/data/online-retail-dataset.csv")
df_retail.show()
```

## Get unique value from column Country

```python
df_retail.select("Country").distinct().show()
```

## Filter data with Country from Denmark

```python
df_retail_denmark = df_retail.filter(F.col("Country") == "Denmark")
df_retail_denmar_count = df_retail_denmark.count()
print(f"\njumlah data dengan filter Negara Denmark :  {df_retail_denmar_count} ")
```

## Cek Harga tertinggi di Negara Denmark per customer ID

```python
df_retail_denmark_max = df_retail_denmark.groupby("CustomerID") \
    .agg(max("UnitPrice").alias("harga_tertinggi")) \
    .orderBy(desc("harga_tertinggi"))

df_retail_denmark_max.show()
```
